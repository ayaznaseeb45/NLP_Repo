Step #1 : Tokenization

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk import word_tokenize
print(word_tokenize("I am reading NLP Fundamentals."))
```

Step 2: Parts-of-Speech (PoS) Tagging

```python
from nltk import word_tokenize, pos_tag
import nltk
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
sentence = "I am reading NLP Fundamentals"
tokens = word_tokenize(sentence)
pos_tags = pos_tag(tokens)
print(pos_tags)
```

Step 3: Stop Word Removal

```python
from nltk.corpus import stopwords
from nltk import word_tokenize
import nltk
nltk.download('stopwords')
stop_words = stopwords.words('english')
sentence = "I am learning Python. It is one of the most popular programming languages."
tokens = word_tokenize(sentence)
def remove_stop_words(tokens, stop_words):
    return [w for w in tokens if w.lower() not in stop_words]
filtered_tokens = remove_stop_words(tokens, stop_words)
print("Original Tokens:", tokens)
print("After Stopword Removal:", filtered_tokens)
```

Step 4: Text Normalization

```python

def normalize(text):
    return text.replace("US", "United States")\
               .replace("UK", "United Kingdom")\
               .replace("-18", "-2018")
sentence = "I visited the US from the UK on 22-10-18"
normalized_text = normalize(sentence)
print("Original:", sentence)
print("Normalized:", normalized_text)
```

Step 5: Spelling Correction

```python
from nltk import word_tokenize
from autocorrect import Speller
spell = Speller(lang='en')
print("Single word correction:", spell("Natureal"))
sentence = "Ntural Luanguage Processin deals with insightes"
tokens = word_tokenize(sentence)
corrected_tokens = [spell(w) for w in tokens]
print("Original Tokens:", tokens)
print("Corrected Tokens:", corrected_tokens)
print("Corrected Sentence:", " ".join(corrected_tokens))
```

Step 6: Stemming

```python
from nltk import stem
porter = stem.PorterStemmer()
words = ["production", "coming", "firing"]
stems = [porter.stem(w) for w in words]
print("Original Words:", words)
print("After Stemming:", stems)
```

Step 7: Lemmatization

```python
import nltk
from nltk.stem.wordnet import WordNetLemmatizer
nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()
words = ["products", "coming"]
lemmas = [lemmatizer.lemmatize(w) for w in words]
print("Original Words:", words)
print("After Lemmatization:", lemmas)
```

Step 8: Named Entity Recognition (NER)

```python
from nltk import word_tokenize, pos_tag, ne_chunk
from nltk import download
download('punkt')
download('averaged_perceptron_tagger')
download('averaged_perceptron_tagger_eng')
download('maxent_ne_chunker')
download('maxent_ne_chunker_tab')
download('words')
sentence = "We are reading a book published by Packt which is based out of Birmingham."
def get_ner(text):
    return [a for a in ne_chunk(pos_tag(word_tokenize(text)), binary=True) if len(a) == 1]
print(get_ner(sentence))
```

Step 9: Word Sense Disambiguation

```python
import nltk
from nltk.wsd import lesk
from nltk import word_tokenize
from nltk import download
download('wordnet')
sentence1 = "Keep your savings in the bank"
sentence2 = "It's so risky to drive over the banks of the road"
def get_synset(sentence, word):
    return lesk(word_tokenize(sentence), word)
print(get_synset(sentence1, 'bank'))
print(get_synset(sentence2, 'bank'))
```

Step 10: Sentence Boundary Detection

```python
import nltk
from nltk.tokenize import sent_tokenize
from nltk import download
download('punkt')
text1 = "We are reading a book. Do you know who is the publisher? It is Packt. Packt is based out of Birmingham."
text2 = "Mr. Donald John Trump is the current president of the USA. Before joining politics, he was a businessman."
def get_sentences(text):
    return sent_tokenize(text)
print(get_sentences(text1))
print(get_sentences(text2))
```
