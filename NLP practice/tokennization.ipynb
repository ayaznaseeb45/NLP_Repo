Tokennization:
Tokenization means breaking text into smaller parts (tokens) — like words, sentences, or symbols — so that a computer can understand and process the text easily.

Text: "I love Python programming."
Tokens: ["I", "love", "Python", "programming", "."]


```python
x = "This is Ayaz"
print(x)
z = x.split()
print(z)
print(z[0])
```

Normalizing case in tokenization 

```python
import re

text = "I'm with you for the       entire life in        U.K.!"
print(text)

# Step 1: Remove extra white spaces using regex
text = re.sub(r'\s+', ' ', text.strip()) 
print(text)

# Step 2: Split sentence into words
words = text.split()
# Step 3: Convert each word to lowercase
words = [word.lower() for word in words]

# Step 4: Print first 100 words (or all if less)
print(words[:100])
```

Spacy libraray use for tokennization

```python
!pip install -q spacy

# 2️ Download English model
!python -m spacy download en_core_web_sm -q

import spacy

# Load English model
nlp = spacy.load("en_core_web_sm")

# 5 Example text
text = "Ayaz is learning Python and NLP."

# 6️ Process text
doc1 = nlp(text)

# 7️ Print tokens and their parts of speech
print("Token -> Part of Speech")
for token in doc1:
    print(f"{token.text} -> {token.pos_}")
```

NER = Name entity recognizer

```python
doc8 = nlp("I have an Apple to build Hong Kong factory for $6 millions ")
for token in doc8:
    print(token.text, end=' | ')
print("\n-------")
for ent in doc8.ents:
    print(ent.text + " " + ent.label_+" - " + str(spacy.explain(ent.label_)))
```

Find Chunks noun

```python
doc9 = nlp("Autonomus cars shift insurance liability towards manufactures.")

for chunks in doc9.noun_chunks:
    print(chunks.text)
```
