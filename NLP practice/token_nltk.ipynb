**NLTK Tokenizer Examples**

Whitespace Tokenizer – Splits text wherever there’s a space, tab, or newline.

Tweet Tokenizer – Cleans and tokenizes tweets, handling special characters and emojis.

MWE Tokenizer – Treats predefined multi-word phrases as a single token.

Regular Expression Tokenizer – Splits text based on a pattern you define using regex.

WordPunctTokenizer – Splits text into words and punctuation as separate tokens.

```python
# Section 1: NLTK Tokenizers
import nltk
from nltk.tokenize import WhitespaceTokenizer, TweetTokenizer, MWETokenizer, RegexpTokenizer, WordPunctTokenizer

# -----------------------------
# Ensure 'punkt' is downloaded in the correct location
nltk.download('punkt', download_dir=r"C:\nltk_data")  # replace path if needed

# Tell NLTK where to look
nltk.data.path.append(r"C:\nltk_data")  # same path as above

# Sample text
text_nltk = "I'm visiting the United States of America! Not only that, I love Python."

# 1️⃣ Whitespace Tokenizer
ws_tokens = WhitespaceTokenizer().tokenize(text_nltk)
print("1. Whitespace Tokenizer:", ws_tokens)

# 2️⃣ Tweet Tokenizer
tweet_tokens = TweetTokenizer().tokenize(text_nltk)
print("2. Tweet Tokenizer:", tweet_tokens)



# 4️⃣ Regex Tokenizer
regex_tokens = RegexpTokenizer(r'\w+').tokenize(text_nltk)
print("4. Regex Tokenizer:", regex_tokens)

# 5️⃣ WordPunctTokenizer
wp_tokens = WordPunctTokenizer().tokenize(text_nltk)
print("5. WordPunctTokenizer:", wp_tokens)
```
