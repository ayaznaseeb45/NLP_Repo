```python
import re
```

```python
text = """
John's email is john.doe@example.com and he will attend the meeting on 23/09/2023. The project code is PRJ-456 and his contact number is +1-234-567-890. Please send the invoice to john.doe@workplace.org.
"""
```

```python
# Regex patterns
email_pattern = r"[A-Za-z0-9._]+@[A-Za-z]+\.[A-Za-z]+"
date_pattern = r"\d{2}/\d{2}/\d{4}"
project_code_pattern = r"[A-Za-z]{3}-\d{3}"
phone_pattern = r"\+\d{1,3}-\d{3}-\d{3}-\d{3}"
```

```python
# Finding matches
emails = re.findall(email_pattern, text)
dates = re.findall(date_pattern, text)
project_codes = re.findall(project_code_pattern, text)
phones = re.findall(phone_pattern, text)
```

```python
print("Emails found:", emails)
print("Dates found:", dates)
print("Project Codes found:", project_codes)
print("Phone Numbers", phones)
```
