```python
import nltk
from nltk import download
```

Step 1: Tokenization

```python
from nltk import word_tokenize

def get_tokens(sentence: str):
    return word_tokenize(sentence)

print(get_tokens("I am reading NLP Fundamentals."))
```

Step 2: Parts-of-Speech (PoS) Tagging 


```python
from nltk import pos_tag

words = get_tokens("I am reading NLP Fundamentals")
print(pos_tag(words))
```

```python
from nltk.corpus import stopwords

stop_words = stopwords.words('english')
sentence = "I am learning Python. It is one of the most popular programming languages."
tokens = word_tokenize(sentence)

def remove_stop_words(tokens, stop_words):
    return [w for w in tokens if w.lower() not in stop_words]

print(remove_stop_words(tokens, stop_words))
```

```python
def normalize(text: str) -> str:
    return (
        text.replace("US", "United States")
            .replace("UK", "United Kingdom")
            .replace("-18", "-2018")
    )

print(normalize("I visited the US from the UK on 22-10-18"))
```

```python
from autocorrect import Speller
spell = Speller(lang='en')
print(spell("Natureal"))

tokens = word_tokenize("Ntural Luanguage Processin deals with insightes")
print([spell(w) for w in tokens])
```

```python
from nltk import stem
porter = stem.PorterStemmer()
print(porter.stem("production"))
print(porter.stem("coming"))
print(porter.stem("firing"))
```

```python
from nltk.stem.wordnet import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
print(lemmatizer.lemmatize("products"))
print(lemmatizer.lemmatize("coming"))
```

```python
from nltk import ne_chunk

sentence = "We are reading a book published by Packt which is based out of Birmingham."

def get_ner(text):
    
    tree = ne_chunk(pos_tag(word_tokenize(text)), binary=True)
    
    return [a for a in tree if len(a) == 1]

print(get_ner(sentence))
```

```python
from nltk.wsd import lesk

sentence1 = "Keep your savings in the bank"
sentence2 = "It's so risky to drive over the banks of the road"

def get_synset(sentence, word):
    return lesk(word_tokenize(sentence), word)

print(get_synset(sentence1, 'bank'))  
print(get_synset(sentence2, 'bank')) 
```

```python
from nltk.tokenize import sent_tokenize

def get_sentences(text: str):
    return sent_tokenize(text)

text1 = (
    "We are reading a book. Do you know who is the publisher? It is Packt. "
    "Packt is based out of Birmingham."
)
text2 = (
    "Mr. Donald John Trump is the current president of the USA. "
    "Before joining politics, he was a businessman."
)

print(get_sentences(text1))
print(get_sentences(text2))
```

```python
ex4_text = "He visited the USA in 19-09-20 and later moved to UK."
def normalize_extended(text: str) -> str:
    return (
        text.replace("USA", "United States of America")
            .replace("US", "United States")
            .replace("UK", "United Kingdom")
            .replace("-20", "-2020")
    )

normalize_extended(ex4_text)
```
